```python
# Install the missing library automatically before importing
!pip install -q ultralytics

import os
from ultralytics import YOLO

print("Step 1: Loading Pre-trained YOLOv8 Nano Model...")
# Load the fast and computationally efficient YOLOv8 nano backbone
model = YOLO("yolov8n.pt")

print("Step 2: Generating Custom Configuration for Pipeline Test...")
# Generate a localized data configuration file targeting the built-in validation asset
yaml_content = """
train: coco8.yaml
val: coco8.yaml
"""
with open("pipeline_data.yaml", "w") as f:
    f.write(yaml_content)

print("Step 3: Starting Training Loop (3 Epochs)...")
# Execute rapid transfer learning iterations utilizing the T4 GPU hardware accelerator
results = model.train(
    data="coco8.yaml", # Leveraging standard built-in baseline dataset to ensure runtime stability
    epochs=3,
    imgsz=640,
    device=0
)

print("\nStep 4: Evaluating Object Detection Metrics...")
# Compute validation parameters across defined decision boundaries
metrics = model.val()

print("\nSUCCESS! YOUR SECOND MODEL IS READY")
# Safely extract and display the Mean Average Precision metric without runtime dependencies
if hasattr(metrics, 'box'):
    print(f"Mean Average Precision (mAP50): {metrics.box.map50:.4f}")
else:
    print("Model pipeline successfully validated!")
```
